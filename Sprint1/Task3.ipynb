```
using System;
using System.Collections.Generic;
using System.Linq;

public class Person
{
    private string _name;
    private int _age;
    private string _gender;

    public string Name
    {
        get => _name;
        set
        {
            if (!string.IsNullOrWhiteSpace(value)) _name = value;
            else throw new ArgumentException("Имя не может быть пустым");
        }
    }

    public int Age
    {
        get => _age;
        set
        {
            if (value >= 0 && value <= 120) _age = value;
            else throw new ArgumentOutOfRangeException("Возраст должен быть от 0 до 120 лет");
        }
    }

    public string Gender
    {
        get => _gender;
        set
        {
            if (value == "Мужской" || value == "Женский") _gender = value;
            else throw new ArgumentException("Пол должен быть 'Мужской' или 'Женский'");
        }
    }

    public Person(string name, int age, string gender)
    {
        Name = name;
        Age = age;
        Gender = gender;
    }

    public virtual string GetInfo()
    {
        return $"Имя: {Name}, Возраст: {Age}, Пол: {Gender}";
    }

    public virtual void SayHello()
    {
        Console.WriteLine($"Привет! Меня зовут {Name}.");
    }
}

public class Student : Person
{
    private string _university;
    private int _course;
    private List<int> _grades = new List<int>();

    public string University
    {
        get => _university;
        set
        {
            if (!string.IsNullOrWhiteSpace(value)) _university = value;
            else throw new ArgumentException("Название университета не может быть пустым");
        }
    }

    public int Course
    {
        get => _course;
        set
        {
            if (value >= 1 && value <= 6) _course = value;
            else throw new ArgumentOutOfRangeException("Курс должен быть от 1 до 6");
        }
    }

    public Student(string name, int age, string gender, string university, int course)
        : base(name, age, gender)
    {
        University = university;
        Course = course;
    }

    public void AddGrade(int grade)
    {
        if (grade >= 2 && grade <= 5) _grades.Add(grade);
    }

    public double GetAverageGrade()
    {
        return _grades.Count > 0 ? _grades.Average() : 0;
    }

    public override void SayHello()
    {
        Console.WriteLine($"Привет! Я студент {Name}, учусь на {Course}-м курсе в {University}.");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Университет: {University}, Курс: {Course}, Средний балл: {GetAverageGrade():F2}";
    }

    public void Study()
    {
        Console.WriteLine($"{Name} сейчас учится.");
    }
}

public class Employee : Person
{
    private string _company;
    private decimal _salary;

    public string Company
    {
        get => _company;
        set
        {
            if (!string.IsNullOrWhiteSpace(value)) _company = value;
            else throw new ArgumentException("Название компании не может быть пустым");
        }
    }

    public decimal Salary
    {
        get => _salary;
        set
        {
            if (value >= 0) _salary = value;
            else throw new ArgumentOutOfRangeException("Зарплата не может быть отрицательной");
        }
    }

    public Employee(string name, int age, string gender, string company, decimal salary)
        : base(name, age, gender)
    {
        Company = company;
        Salary = salary;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Компания: {Company}, Зарплата: {Salary} руб.";
    }

    public void Work()
    {
        Console.WriteLine($"{Name} сейчас работает в {Company}.");
    }
}

public class Teacher : Person
{
    private string _subject;
    private int _experience;
    private List<Student> _students = new List<Student>();

    public string Subject
    {
        get => _subject;
        set
        {
            if (!string.IsNullOrWhiteSpace(value)) _subject = value;
            else throw new ArgumentException("Название предмета не может быть пустым");
        }
    }

    public int Experience
    {
        get => _experience;
        set
        {
            if (value >= 0 && value <= 70) _experience = value;
            else throw new ArgumentOutOfRangeException("Стаж должен быть от 0 до 70 лет");
        }
    }

    public Teacher(string name, int age, string gender, string subject, int experience)
        : base(name, age, gender)
    {
        Subject = subject;
        Experience = experience;
    }

    public void AddStudent(Student student)
    {
        if (!_students.Contains(student))
        {
            _students.Add(student);
            Console.WriteLine($"Студент {student.Name} добавлен к преподавателю {Name}");
        }
    }

    public void GradeStudent(Student student, int grade)
    {
        if (_students.Contains(student))
        {
            student.AddGrade(grade);
            Console.WriteLine($"Преподаватель {Name} выставил оценку {grade} студенту {student.Name} по предмету {Subject}");
        }
    }

    public override void SayHello()
    {
        Console.WriteLine($"Здравствуйте! Я преподаватель {Name}, веду предмет '{Subject}'.");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Предмет: {Subject}, Стаж: {Experience} лет, Студентов: {_students.Count}";
    }

    public void Teach()
    {
        Console.WriteLine($"{Name} преподаёт {Subject}.");
    }
}

public class University
{
    public string Name { get; set; }
    private List<Student> _students = new List<Student>();
    private List<Teacher> _teachers = new List<Teacher>();
    private List<Employee> _employees = new List<Employee>();

    public University(string name)
    {
        Name = name;
    }

    public void AddStudent(Student student) => _students.Add(student);
    public void AddTeacher(Teacher teacher) => _teachers.Add(teacher);
    public void AddEmployee(Employee employee) => _employees.Add(employee);

    public void OrganizeLearning()
    {
        Console.WriteLine($"Учебный процесс в {Name}:");
        for (int i = 0; i < _students.Count; i++)
        {
            var teacher = _teachers[i % _teachers.Count];
            teacher.AddStudent(_students[i]);
            teacher.GradeStudent(_students[i], 5);
        }
    }

    public void PaySalaries()
    {
        Console.WriteLine($"\nВыплата зарплат в {Name}:");
        foreach (var e in _employees)
        {
            Console.WriteLine($"Сотрудник {e.Name} получил зарплату {e.Salary} руб.");
        }
    }
}

var student1 = new Student("Анна Смирнова", 20, "Женский", "ТюмГУ", 3);
var student2 = new Student("Олег Новиков", 22, "Мужской", "ТюмГУ", 4);
var employee = new Employee("Пётр Сидоров", 35, "Мужской", "ТюмГУ", 75000);
var teacher1 = new Teacher("Елена Иванова", 45, "Женский", "Математика", 15);
var teacher2 = new Teacher("Дмитрий Соколов", 50, "Мужской", "Физика", 25);

var uni = new University("ТюмГУ");
uni.AddStudent(student1);
uni.AddStudent(student2);
uni.AddTeacher(teacher1);
uni.AddTeacher(teacher2);
uni.AddEmployee(employee);

uni.OrganizeLearning();
uni.PaySalaries();

student1.Course = 4;
employee.Salary = 85000;

Console.WriteLine("\nИтоговая информация:");
Console.WriteLine(student1.GetInfo());
Console.WriteLine(student2.GetInfo());
Console.WriteLine(employee.GetInfo());
Console.WriteLine(teacher1.GetInfo());
```
